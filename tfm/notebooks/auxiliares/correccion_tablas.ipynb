Errores detectados a la hora de improtar los datos a las tablas de SQL

1. Habia valores de la tabla de pharmacy (columna one_hr_max) que tenian un formato que no detectaban

Solucion:
    1. Identificar esos registros
    2. Unir tablas pharmacy y prescriptions por el pharmacy_id para ver a que medicamento correspondian esos registros.
    

```python
import pandas as pd
```

```python
pharmacy=pd.read_csv(r'DATA_ROOT\mimic-iv-3.1\mimic-iv-3.1\hosp\pharmacy.csv\pharmacy.csv')
prescription=pd.read_csv(r"DATA_ROOT\mimic-iv-3.1\mimic-iv-3.1\hosp\prescriptions.csv\prescriptions_corregido.csv")
```

```python
# Buscar valores con formato incorrecto
invalid_values = pharmacy[pharmacy['one_hr_max'].str.contains(r'\d+-\d+', na=False)]
print(invalid_values)

# Mostrar los registros con valores mal formateados
print("Registros con valores mal formateados en 'one_hr_max':")
print(invalid_values[['pharmacy_id', 'one_hr_max']])

```

```python
# Eliminar las columnas no deseadas
pharmacy= pharmacy.drop(columns=["sliding_scale", "fill_quantity", "lockout_interval", "basal_rate", "expirationdate"])
```

```python
pharmacy=pharmacy.drop(columns=['one_hr_max'])
```

```python
pharmacy.info()
```

```python
# Guardar el archivo sin esas columnas
pharmacy.to_csv(r'DATA_ROOT\mimic-iv-3.1\mimic-iv-3.1\hosp\pharmacy.csv\pharmacy_cleaned.csv', index=False)
```

```python
poe=pd.read_csv(r"DATA_ROOT\mimic-iv-3.1\mimic-iv-3.1\hosp\poe.csv\poe.csv")
```

```python
poe.info()
```

```python
# Ver los valores únicos en ordertime
print(poe["ordertime"].unique()[:50])  # Mostrar 50 valores únicos
```

```python
# Filtrar valores incorrectos (que terminan en ":" en lugar de ":SS")
invalid_dates = poe[poe["ordertime"].str.match(r'^\d{4}-\d{2}-\d{2} \d{2}:\d{2}:?$', na=False)]

# Mostrar los primeros valores problemáticos
print(invalid_dates["ordertime"].head(20))
```

```python
# Buscar valores numéricos inesperados
numeric_errors = poe[poe["ordertime"].str.match(r'^\d+$', na=False)]
print(numeric_errors["ordertime"].head(20))
```

```python
# Filtrar y eliminar el registro donde ordertime es exactamente "218"
poe = poe[poe["ordertime"] != "218"]
```

```python
poe=poe.drop(columns=["discontinue_of_poe_id", "discontinued_by_poe_id", "transaction_type"])
```
