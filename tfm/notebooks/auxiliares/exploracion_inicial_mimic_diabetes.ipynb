```python
def info_con_valores_unicos(df):
    """
    Imprime la información del DataFrame (df.info()) junto con 
    los valores únicos de cada variable.
    """
    # Mostrar información básica del DataFrame
    df.info()

    # Separador visual
    print("\n" + "="*50 + "\n")

    # Mostrar valores únicos de cada variable
    for col in df.columns:
        print(f"Variable: {col}")
        print(f" - Tipo de dato: {df[col].dtype}")
        print(f" - Valores únicos ({df[col].nunique()}): {df[col].unique().tolist()}")
        print("-" * 50)
```
