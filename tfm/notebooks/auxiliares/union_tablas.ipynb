## Crear subconjunto con pacientes diabeticos tipo 1

1. Unir tablas diagnoses y admissions

```python
# Unir dos tablas por la columna 'subject_id'
# Verificar si ambas tablas contienen la columna 'subject_id'
if 'subject_id' in admissions.columns and 'subject_id' in diagnoses.columns:
    admin_diag = diagnoses.merge(admissions, on='subject_id', how='inner')  # INNER JOIN por 'subject_id'
admin_diag.head()    
```

2. Filtrar por los codigos ICD-9, ICD-10

```python
# Filtrar ICD-9: Códigos que empiezan con '250' y tienen el quinto dígito '1'
admin_diag_icd9_tipo1 = admin_diag[admin_diag["icd_code"].str.match(r"^250\d1$", na=False)]

# Filtrar ICD-10: Códigos que empiezan con 'E10'
admin_diag_icd10_tipo1 = admin_diag[admin_diag["icd_code"].str.startswith("E10", na=False)]

# Combinar ambos filtros en un solo DataFrame
admin_diag_diabetes_tipo1 = pd.concat([admin_diag_icd9_tipo1, admin_diag_icd10_tipo1]).drop_duplicates()

# Mostrar resultados
print(f"Número de pacientes únicos con diabetes tipo 1: {admin_diag_diabetes_tipo1['subject_id'].nunique()}")
admin_diag_diabetes_tipo1.head()
```
