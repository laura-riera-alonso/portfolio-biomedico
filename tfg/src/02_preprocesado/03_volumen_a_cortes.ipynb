```python
from skimage.transform import rescale, resize
from skimage.io import imshow,imread,show,imsave
import numpy as np
from pathlib import Path
import scipy.io
import os 
import nibabel as nib
```

```python
#SCRIPT TO GET THE SLICES OF EVERY cropped VOLUME
input_folder=r'DATA_ROOT/preprocesado_results_v2'
output_folder=r'DATA_ROOT/volume_slices_all'
mod_list=['T1W_CE_3D_FFE_TRA_IQ','T1W_3D_FFE_TRA_IQ', 'FLAIR_3D_SPIR_TRA_IQ', 'SWIm']
template='T1W_CE_3D_FFE_TRA_IQ'
content_row = ['PAT_'+ str(pat+1) for pat in range(0,20)]

for pat in range(1,122):
    for mod in mod_list:
        print(mod)
        pat_number='PAT_'+str(pat)+'_'
        print(pat_number)
        if not(os.path.isdir(os.path.join(output_folder,mod,pat_number))):
            os.makedirs(os.path.join(output_folder,mod,pat_number))
        im_path=os.path.join(input_folder,mod,pat_number, pat_number+mod+'_final_crop.nii.gz')
        im=nib.load(im_path)
        im_data=im.get_fdata()
        size=im_data.shape
        n_slices=size[2]
        print(n_slices)
        
        for slice in range(0,n_slices):
            im_slice=im_data[:,:,slice]
            #print(im_slice.shape)
            nifti_img=nib.Nifti1Image(im_slice, im.affine, im.header)
            nib.save(nifti_img,os.path.join(output_folder,mod,pat_number,pat_number+mod+'_slice_'+str(slice)+'.nii.gz'))
            
```
